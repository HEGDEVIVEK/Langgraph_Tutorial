```python
import os
from dotenv import load_dotenv
load_dotenv()

if os.environ.get("OPENAI_API_KEY"):
    print("bro it is working")


from langchain.chat_models import init_chat_model
from langchain.tools import tool
from langchain_core.prompts import ChatPromptTemplate

llm_model = init_chat_model(model="gpt-5-nano")
```

```python
from langchain_community.tools import DuckDuckGoSearchRun

@tool
def duckduck_tool(query : str):

    """This is search tool of DuckDuckGo for given quesry and return the results"""
    custom_duckduck_tool = DuckDuckGoSearchRun()
    return  custom_duckduck_tool.invoke(query)
```

```python
from langchain_community.utilities import ArxivAPIWrapper
from langchain_community.tools import ArxivQueryRun

@tool
def arXiv_tool(query: str):

    """This tool allows you to query the arXiv database for research papers"""
    custom_arxiv_tool = ArxivQueryRun(api_wrapper=ArxivAPIWrapper(top_k_results=1))
    return custom_arxiv_tool.invoke(query)
```

```python
@tool
def personal_info(name: str):

    """ use this tool to get the personal information about someone """

    info = {
        "Vivek" : "He is persuing his MS in CS at UTSA",
        "Vikram" : "He is working in Banglore a loan and mortgage analyst"
    }

    return info.get(name, "No information available for this person")
```

TOOL BINDING 

```python
toolKit = [duckduck_tool, arXiv_tool,personal_info ]

llm_with_toolKit = llm_model.bind_tools(toolKit)
```

```python
response = llm_with_toolKit.invoke( "Search arXiv for the research paper 'Attention Is All You Need'.")
response.tool_calls
```
