```python
import os
from dotenv import load_dotenv
load_dotenv()

if os.environ.get("OPENAI_API_KEY"):
    print("bro it is working")


from langchain.chat_models import init_chat_model

llm_model = init_chat_model(model="gpt-5-nano")
```

FIRST GRAPH 

step -1 Define the schema

```python
from typing import TypedDict, List

class graph_schema(TypedDict):
    name : str
    message : str
```

step -2 Create the Node Function 

```python
def welcome(schema:graph_schema):
 
    curr_name = schema["name"]

    response = llm_model.invoke(f"my name is {curr_name}").content

    schema["message"] = f"Here is my response : {response}"

    return schema
```

step -3 Create the state graph

```python
from langgraph.graph import StateGraph, START, END

graph = StateGraph(graph_schema)

graph.add_node("welcome", welcome)

graph.add_edge(START, "welcome")
graph.add_edge("welcome", END)
```

```python
graph = graph.compile()

graph.invoke({
    "name" : "Vivek",
    "message": ""
    })
```
